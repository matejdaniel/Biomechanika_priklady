````{exercise} Veletoč na hrazdě
:label: ex-veletoc

Gymnasta provádí na hrazdě prvek, při kterém provede $2{,}5$ veletoče. Určete velikost úhlu, který opíše jeho těžiště kolem hrazdy ve stupních a radiánech.

```{figure} images/pr-001.png
:width: 250px
:align: center
:name: fig-veletoc

Schéma fázového rozpadu veletoče na hrazdě. Převzato z [Burton a kol, 2023](https://doi.org/10.1080/02640414.2023.2259201).
````

````{solution} ex-veletoc
:class: dropdown
Jeden kompletní velotoč je 360° nebo $2 \pi$ radiánu.

$$2,5 \times 360°= 900˚$$

$$2,5 \times 2 \pi = 6 \pi$$
````


````{exercise} Úhlový rozsah pohybu předloktí
:label: ex-predlokti-sipka

Průměrná úhlová rychlost předloktí při hodu šipkou je $\omega = \frac{\pi}{2}\text{ rad}\cdot\text{s}^{-1}$. Doba trvání pohybu je $t = 0{,}8\text{ s}$. Určete celkový rozsah pohybu předloktí při hodu ve stupních.

```{figure} images/pr-007.png
:width: 250px
:align: center
:name: fig-predlokti-sipka

Kinematika hodu šipkou.
```

````


```{solution} ex-predlokti-sipka
:class: dropdown

Úhlová dráha v radiánech:
$$\Delta\theta = \omega \cdot t = \frac{\pi}{2} \cdot 0{,}8 = 0{,}4\pi\text{ rad}$$

Převod na stupně:
$$\Delta\theta = 0{,}4\pi \cdot \frac{180^\circ}{\pi} = 72^\circ$$

```


````{exercise} Určení zrychlení
:label: ex-rychlost

Na grafu je závislost rychlosti a času. Ve kterém úseku je nejvyšší zrychlení

```{figure} images/prx-001.svg
:width: 450px
:align: center
:name: fig-rychost

Závislost rychlosti na času.
```
````

```{solution} ex-rychlost
:class: dropdown

Zrychlení je dáno směrnicí tečny (sklonem přímky) v grafu rychlosti v závislosti na čase: $a = \frac{\Delta v}{\Delta t}$.

Hodnoty zrychlení v jednotlivých úsecích:

* **0 až 10 min:**

$$a = \frac{60 - 0}{10 - 0} = 6\text{ m/min}^2 = 0{,}1\text{ m/s}^2$$


* **10 až 15 min:**

$$a = \frac{60 - 60}{5} = 0\text{ m/min}^2$$


* **15 až 40 min:**

$$a = \frac{-40 - 60}{40 - 15} = \frac{-100}{25} = -4\text{ m/min}^2$$



(velikost/zpomalení je $4\text{ m/min}^2$)
* **40 až 55 min:**

$$a = \frac{0 - (-40)}{55 - 40} = \frac{40}{15} \approx 2{,}67\text{ m/min}^2$$



Nejvyšší zrychlení (jak algebraicky, tak co do absolutní velikosti) je v **prvním úseku od 0 do 10 minut**, kde činí **$6\text{ m/min}^2$** ($0{,}1\text{ m/s}^2$).

```

````{exercise} Délka dvoukroku
:label: ex-dvoukrok


Běžec provede 5 dvojkroků za 3 s průměrnou rychlostí v = 6 m.s⁻¹. Jaká je průměrná délka jednoho dvojkroku?


```{figure} images/prx-020.png
:width: 250px
:align: center
:name: fig-dvoukrok

Délka dvoukroku.
```
````

```{solution} ex-dvoukrok
:class: dropdown

1. **Výpočet celkové dráhy (s):**
* s = v * t
* s = 6 * 3 = 18 m


2. **Výpočet délky jednoho dvojkroku (d):**
* d = s / n
* d = 18 / 5 = 3,6 m

**Délka dvojkroku je 3,6 m.**

```

````{exercise} Rychlost kolene při rotaci
:label: ex-koleno

**Zadání:**

Cyklista jede na kole tak, že jeho stehno rotuje rychlostí $\omega_s = 60^{\circ}\cdot\text{s}^{-1}$ (extenze kyčle) a úhlové zrychlení směrem do flexe v koleni je $\alpha_k = 5^{\circ}\cdot\text{s}^{-2}$.

Určete:

1. Obvodovou rychlost kolena vůči kyčli $v_k$, pokud víte, že délka stehna $d_s$ je 0,5 m (50 cm).
2. Obvodové zrychlení bérce vůči kolenu $a_b$, pokud víte, že délka bérce $d_b$ je 0,4 m (40 cm).

Předpokládejte, že koleno je bod na stehně. 

```{figure} images/prx-021.png
:width: 350px
:align: center
:name: fig-koleno

Jízda na kole
```
````

```{solution} ex-koleno
:class: dropdown

Zadané údaje:

* $\omega_s = 60^{\circ}\cdot\text{s}^{-1}$ (úhlová rychlost stehna)
* $\alpha_k = 5^{\circ}\cdot\text{s}^{-2}$ (úhlové zrychlení kolene)
* $d_s = 0,5\text{ m} = 50\text{ cm}$ (délka stehna)
* $d_b = 0,4\text{ m} = 40\text{ cm}$ (délka bérce)

**1. Převod úhlů na radiány:**

Pro výpočet musíme převést úhlové veličiny ze stupňů na radiány.
Víme, že $180^{\circ} = \pi\text{ rad}$.

* $\omega_s = \frac{60 \cdot \pi}{180} = \frac{\pi}{3}\text{ rad}\cdot\text{s}^{-1} \approx 1,05\text{ rad}\cdot\text{s}^{-1}$
* $\alpha_k = \frac{5 \cdot \pi}{180} = \frac{\pi}{36}\text{ rad}\cdot\text{s}^{-2} \approx 0,087\text{ rad}\cdot\text{s}^{-2}$

**2. Výpočet rychlosti kolene ($v_k$):**
Předpokládáme, že koleno je bod na stehně. Obvodová rychlost bodu je dána vztahem $v = \omega \cdot d$.

* $v_k = \omega_s \cdot d_s$
* $v_k = 1,05 \cdot 0,5 = 0,525\text{ m}\cdot\text{s}^{-1}$

**3. Výpočet obvodového zrychlení bérce ($a_b$):**
Obvodové zrychlení bodu je dáno vztahem $a = \alpha \cdot d$.

* $a_b = \alpha_k \cdot d_b$
* $a_b = 0,087 \cdot 0,4 = 0,035\text{ m}\cdot\text{s}^{-2}$

**Odpovědi:**

1. Obvodová rychlost kolena je **$0,53\text{ m}\cdot\text{s}^{-1}$**.
2. Obvodové zrychlení bérce je **$0,035\text{ m}\cdot\text{s}^{-2}$**.


```
