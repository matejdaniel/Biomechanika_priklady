# Síly v biomechanice

Cílem této kapitoly je opakovat principy sil, jejich rozložení na složky a zároveň jejich skládání. 

````{exercise} Odraz běžce
:label: ex-odraz-bezce

Běžec se odráží silou $F = 2000\text{ N}$ tak, že vektor síly svírá s horizontálou úhel $\alpha = 60^\circ$. Určete složky vektoru odrazové síly v horizontálním a vertikálním směru.

```{figure} images/pr-000.png
:width: 250px
:align: center
:name: fig-odraz-bezce

Schéma odrazové síly běžce.

```

````

````{solution} ex-odraz-bezce
:class: dropdown

Horizontální složka síly:
$$F_x = F \cdot \cos\alpha = 2000 \text{ N}\cdot \cos(60^\circ) = 2000 \text{ N}\cdot 0{,}5 = 1000\text{ N}$$

Vertikální složka síly:
$$F_y = F \cdot \sin\alpha = 2000 \text{ N}\cdot \sin(60^\circ) = 2000 \text{ N}\cdot \frac{\sqrt{3}}{2} \approx 1732{,}05\text{ N}$$

````


````{exercise} Tření podrážky běžce
:label: ex-treni-podrazky

Koeficient statického tření mezi podrážkou boty běžce a zemí je $0{,}5$ a koeficient dynamického tření je $0{,}3$. 
    Působí-li běžec na zem vertikální sílou $2000\text{ N}$ (*ground reaction force*), 
    určete maximální horizontální sílu $F_T$ , kterou může běžec vyvinout bez podklouznutí nohy. Vysvětlete směr jednotlivých sil. 

```{figure} images/pr-004.png
:width: 250px
:align: center
:name: fig-treni-podrazky

Působení reakčních sil na chodidlo.
```

````


```{solution} ex-treni-podrazky
:class: dropdown

Podmínka pro mezní stav bez podklouznutí je dána maximální statickou třecí silou:

$$F_{T,\max} = f_s \cdot F_N = 0{,}5 \cdot 2000\text{ N} = 1000\text{ N}$$

Z fyzikálního a biomechanického hlediska působí síly na běžce následujícím způsobem:

* **$F_W$ (Tíhová síla / Weight):** Směřuje svisle dolů. Zastupuje celkovou tíhu těla a působí v jeho těžišti (Center of Gravity). Je to síla, kterou je těleso přitahováno k Zemi.
* **$F_d$ (Odpor vzduchu / Drag force):** Směřuje vodorovně proti směru pohybu běžce. Představuje aerodynamický odpor, kterým okolní vzduch brání tělu v dopředném pohybu.
* **$F_N$ (Normálová síla / Normal force):** Směřuje svisle vzhůru z místa kontaktu chodidla s podložkou. Jedná se o kolmou reakci povrchu (Ground Reaction Force) na tlak, kterým noha působí na zem. Zabraňuje propadnutí běžce do dráhy.
* **$F_T$ (Třecí síla / Tractive, Friction force):** Směřuje vodorovně vpřed, tedy ve směru pohybu. V této fázi odrazu (push-off) funguje jako hnací síla. Když běžec působí nohou na dráhu silou směřující dozadu, dráha (díky statickému tření a 3. Newtonovu zákonu akce a reakce) působí na chodidlo stejně velkou silou směřující dopředu. Horizontální složka této reakce je právě $F_T$, která zajišťuje dopředný pohyb.

```


````{exercise} Skládání dvojice sil
:label: ex-skladani-dvojice-sil

Na těleso působí dvojice sil. Síly jsou vodorovně nebo pod úhlem $45^\circ$. Určete, ve kterém případě bude výslednice sil maximální.

```{figure} images/pr-006.png
:width: 250px
:align: center
:name: fig-skladani-dvojice-sil

Konfigurace vektorů sil.
```

````


```{solution} ex-skladani-dvojice-sil
:class: dropdown

Pro výpočet výslednice sil v jednotlivých variantách budeme vycházet z předpokladu (který je zřejmý z vizuálního zobrazení v předloze), že šikmé vektory svírají s vodorovnou osou úhel **45°**.

Síly budeme sčítat vektorově. Každou sílu si můžeme rozložit na horizontální (osu $x$) a vertikální (osu $y$) složku.

### Společný vektor

Ve všech čtyřech variantách vystupuje stejný první vektor (mířící vpravo nahoru):

* Velikost: 10 N
* Složka $x$: $F_{1x} = 10 \cdot \cos(45^\circ) = 5\sqrt{2} \doteq 7{,}07\text{ N}$
* Složka $y$: $F_{1y} = 10 \cdot \sin(45^\circ) = 5\sqrt{2} \doteq 7{,}07\text{ N}$
* Vektorový zápis: $\vec{F_1} = (7{,}07; 7{,}07)\text{ N}$

---

### Varianta A

Druhý vektor míří vlevo nahoru pod úhlem 45°. Zrcadlově se tak odečte horizontální složka a sečte vertikální.

* Druhý vektor: $\vec{F_2} = (-10 \cdot \cos(45^\circ); 10 \cdot \sin(45^\circ)) = (-7{,}07; 7{,}07)\text{ N}$
* Výslednice (součet složek):

$$\vec{R_A} = (7{,}07 - 7{,}07; 7{,}07 + 7{,}07) = (0; 14{,}14)\text{ N}$$


* **Velikost výslednice:** $R_A = \mathbf{14{,}14\text{ N}}$ (výsledný vektor směřuje kolmo vzhůru)

### Varianta B

Druhý vektor míří vodorovně doleva, má tedy pouze zápornou složku $x$.

* Druhý vektor: $\vec{F_2} = (-10; 0)\text{ N}$
* Výslednice (součet složek):

$$\vec{R_B} = (7{,}07 - 10; 7{,}07 + 0) = (-2{,}93; 7{,}07)\text{ N}$$


* **Velikost výslednice:** Pomocí Pythagorovy věty $R_B = \sqrt{(-2{,}93)^2 + 7{,}07^2} = \sqrt{8{,}58 + 49{,}98} \doteq \mathbf{7{,}65\text{ N}}$

### Varianta C

Druhý vektor míří vodorovně doprava s poloviční velikostí.

* Druhý vektor: $\vec{F_2} = (5; 0)\text{ N}$
* Výslednice (součet složek):

$$\vec{R_C} = (7{,}07 + 5; 7{,}07 + 0) = (12{,}07; 7{,}07)\text{ N}$$


* **Velikost výslednice:** $R_C = \sqrt{12{,}07^2 + 7{,}07^2} = \sqrt{145{,}68 + 49{,}98} \doteq \mathbf{13{,}99\text{ N}}$

### Varianta D

Druhý vektor míří vodorovně doleva s poloviční velikostí.

* Druhý vektor: $\vec{F_2} = (-5; 0)\text{ N}$
* Výslednice (součet složek):

$$\vec{R_D} = (7{,}07 - 5; 7{,}07 + 0) = (2{,}07; 7{,}07)\text{ N}$$


* **Velikost výslednice:** $R_D = \sqrt{2{,}07^2 + 7{,}07^2} = \sqrt{4{,}28 + 49{,}98} \doteq \mathbf{7{,}37\text{ N}}$


```


````{exercise} Hmotnost dítěte na houpačce
:label: ex-dite-houpacka

Dítě je zavěšeno v houpačce podle obrázku. Určete hmotnost dítěte.

```{figure} images/pr-011.jpg
:width: 250px
:align: center
:name: fig-dite-houpacka

Statické silové schéma závěsu houpačky.
```
````

```{solution} ex-dite-houpacka
:class: dropdown

**Podmínka rovnováhy ve svislém směru:**

Horizontální složky tahových sil v lanech se díky symetrii navzájem vyruší. Celkovou tíhovou sílu dítěte $F_G$ vyrovnávají vertikální složky obou tahových sil:

$$F_G = 2 \cdot F \cdot \cos(40^\circ)$$

Dosazení zadaných hodnot ($F = 50\text{ N}$):

$$F_G = 2 \cdot 50 \cdot \cos(40^\circ) = 100 \cdot 0{,}7660 \approx 76{,}6\text{ N}$$

**Výpočet hmotnosti:**

$$m = \frac{F_G}{g} = \frac{76{,}6}{9{,}81} \approx 7{,}81\text{ kg}$$

```

````{exercise} Statická rovnováha knihy
:label: ex-kniha-salek

Kniha o hmotnosti $m = 1\text{ kg}$ je spojena provázkem se zavěšeným šálkem o hmotnosti 100 g. Koeficient statického tření mezi knihou a podložkou je $f_s = 0{,}5$. Jaký maximální objem vody můžeme nalít do šálku, aby se kniha nezačala pohybovat (rychlost $v$ nulová)? 

```{figure} images/pr-014.jpg
:width: 250px
:align: center
:name: fig-kniha-salek

Soustava těles spojených vláknem.
```

````

````{solution} ex-kniha-salek
:class: dropdown

**Fyzikální rozbor**

Na knihu o hmotnosti $m = 1\text{ kg}$ na nakloněné rovině s úhlem $\alpha = 20^\circ$ působí:

* Tíhová síla rozložená na dvě složky:
* Kolmá složka: $F_n = m \cdot g \cdot \cos(20^\circ)$
* Rovnoběžná složka se sklonem dolů: $F_\parallel = m \cdot g \cdot \sin(20^\circ)$

```{image} images/pr-014s.svg
:width: 250px
:align: center
:name: fig-kniha-salek

Soustava těles spojených vláknem.
```


* Maximální síla statického tření: $F_{t,\text{max}} = f_s \cdot F_n = f_s \cdot m \cdot g \cdot \cos(20^\circ)$
* Tahová síla provázku $T$, která odpovídá tíze zavěšeného šálku $G_{\text{šálek}}$

**Výpočet jednotlivých silových složek**

Při $g = 9{,}81\text{ m}\cdot\text{s}^{-2}$:

* $\cos(20^\circ) \approx 0{,}9397$
* $\sin(20^\circ) \approx 0{,}3420$
* $F_\parallel = 1 \cdot 9{,}81 \cdot 0{,}3420 = 3{,}355\text{ N}$
* $F_n = 1 \cdot 9{,}81 \cdot 0{,}9397 = 9{,}218\text{ N}$
* $F_{t,\text{max}} = 0{,}5 \cdot 9{,}218 = 4{,}609\text{ N}$

**Stanovení maximální tíhy šálku**

Lanko i složka tíže knihy táhnou knihu dolů po nakloněné rovině směrem ke kladce. Proti nim působí statická třecí síla nahoru:

$$T_{\text{max}} + F_\parallel = F_{t,\text{max}}$$

$$M_{\text{max}} \cdot g + m \cdot g \cdot \sin(20^\circ) = f_s \cdot m \cdot g \cdot \cos(20^\circ)$$

Po vykrácení $g$ dostáváme celkovou maximální hmotnost zavěšené zátěže (šálek + voda):

$$M_{\text{max}} = m \cdot \left[ f_s \cos(20^\circ) - \sin(20^\circ) \right]$$

**Číselný výpočet:**

* $\cos(20^\circ) \approx 0{,}9397$
* $\sin(20^\circ) \approx 0{,}3420$

$$M_{\text{max}} = 1 \cdot \left[ 0{,}5 \cdot 0{,}9397 - 0{,}3420 \right] = 0{,}4698 - 0{,}3420 = 0{,}1278\text{ kg} \approx 127{,}8\text{ g}$$

**Objem vody:**

* Hmotnost samotného prázdného šálku: $m_{\text{š}} = 100\text{ g}$
* Maximální hmotnost vody:

$$m_{\text{voda}} = M_{\text{max}} - m_{\text{š}} = 127{,}8\text{ g} - 100\text{ g} = 27{,}8\text{ g}$$



Při hustotě vody $\rho = 1\text{ g}\cdot\text{cm}^{-3}$ je maximální objem:

$$V_{\text{max}} = \frac{m_{\text{voda}}}{\rho} \approx \mathbf{28\text{ ml}}$$

````


````{exercise} Síly na femur
:label: ex-sily-femur

Při trakci působí na femur síly podle obrázku. Určete velikost x-sové složky výslednice.

```{figure} images/prx-006.png
:width: 350px
:align: center
:name: fig-sily-femur

Síly působící na femur při trakci.
```

````

````{solution} ex-sily-femur
:class: dropdown

Pro určení $x$-ové složky výslednice ($R_x$) sečteme vodorovné průměty všech tří sil působících v uzlu:

$$R_x = \sum F_{ix} = F_1 \cos(\theta_1) + F_2 \cos(\theta_2) + F_3 \cos(\theta_3)$$

**Zadané hodnoty:**

* $F_1 = 10\text{ N}$, $\theta_1 = 61^\circ$
* $F_2 = 10\text{ N}$, $\theta_2 = 17^\circ$
* $F_3 = 10\text{ N}$, $\theta_3 = -19^\circ$

**Výpočet:**

$$R_x = 10 \cdot \cos(61^\circ) + 10 \cdot \cos(17^\circ) + 10 \cdot \cos(-19^\circ)$$

Hodnoty goniometrických funkcí (s vědomím, že $\cos(-\alpha) = \cos\alpha$):

* $\cos(61^\circ) \approx 0{,}4848$
* $\cos(17^\circ) \approx 0{,}9563$
* $\cos(-19^\circ) \approx 0{,}9455$

$$R_x = 10 \cdot (0{,}4848 + 0{,}9563 + 0{,}9455) = 10 \cdot 2{,}3866 \approx 23{,}87\text{ N}$$

Velikost $x$-ové složky výslednice tahových sil je přibližně **$23{,}9\text{ N}$** (resp. **$23{,}87\text{ N}$**).

````


````{exercise} Síla na hlavici femuru
:label: ex-sily-hlavice

Na hlavici femuru působí síla 2 kN pod úhlem 60° k podélné ose krčku femuru. Určete složku síly ve směru krčku femuru. 

```{figure} images/prx-011.png
:width: 350px
:align: center
:name: fig-sily-hlavice

Síly působící na femur při trakci.
```

````

````{solution} ex-sily-hlavice
:class: dropdown

**Zadané hodnoty:**

* Celková síla: $F = 2\text{ kN} = 2000\text{ N}$
* Úhel sevřený se směrem osy krčku femuru: $\alpha = 60^\circ$

---

**Výpočet:**

Složka síly působící ve směru podélné osy krčku femuru (osová/tlaková složka $F_\parallel$) je průmětem celkové síly do této osy:

$$F_\parallel = F \cdot \cos(\alpha)$$

Dosazení číselných hodnot:

$$F_\parallel = 2000 \cdot \cos(60^\circ) = 2000 \cdot 0{,}5 = 1000\text{ N} = 1\text{ kN}$$

---

Složka síly ve směru osy krčku femuru je **$1\text{ kN}$** ($1000\text{ N}$). Jedná se o sílu namáhající krček femuru čistým tlakem.

````


````{exercise} Ortodoncie
:label: ex-sily-rovnatka

**Zadání:**

V ortodoncii se k nápravě polohy vystupujícího zubu používá předepjatý drát rovnátka. Na zub působí v místech ohybu drátu dvě tahové síly o stejné velikosti $T = T' = 30\text{ N}$, přičemž obě svírají s tečnou rovinou (vodorovnou referenční přímkou) úhel $\theta = 16{,}0^\circ$ (viz obr.).

**Určete velikost a směr výsledné síly působící na zub.**

---

```{figure} images/prx-012.png
:width: 350px
:align: center
:name: fig-sily-rovnatka

Síly působící na zub při ortodonci.
```

````

````{solution} ex-sily-hlavice
:class: dropdown

Zvolíme kartézský souřadnicový systém:

* Osa $x$ leží ve vodorovném směru (podél referenční přímky).
* Osa $y$ směřuje svisle dolů (dovnitř zubního oblouku, směrem k patru).

**1. Rozklad sil do jednotlivých složek:**

* Pro sílu $\vec{T}$ (vpravo):

$$T_x = T \cdot \cos(16{,}0^\circ), \quad T_y = T \cdot \sin(16{,}0^\circ)$$


* Pro sílu $\vec{T}'$ (vlevo):

$$T'_x = -T \cdot \cos(16{,}0^\circ), \quad T'_y = T \cdot \sin(16{,}0^\circ)$$



**2. Složení sil:**

* **Vodorovná složka výslednice:**
Vzhledem k symetrii se vodorovné složky vzájemně vyruší:

$$F_{Rx} = T_x + T'_x = T \cdot \cos(16{,}0^\circ) - T \cdot \cos(16{,}0^\circ) = 0\text{ N}$$


* **Svislá složka výslednice:**
Svislé složky míří shodně dolů (do středu oblouku) a sčítají se:

$$F_{Ry} = T_y + T'_y = 2 \cdot T \cdot \sin(16{,}0^\circ)$$



**3. Číselný výpočet:**

$$F_R = 2 \cdot 30 \cdot \sin(16{,}0^\circ) = 60 \cdot 0{,}2756 \doteq 16{,}54\text{ N}$$

---

Výsledná síla působící na zub má velikost přibližně **$16{,}5\text{ N}$** a směřuje kolmo k zubnímu oblouku dovnitř (směrem do ústní dutiny / k patru), čímž zub zasouvá do požadované linie. 

````


````{exercise} Délka bérce
:label: ex-sily-drep

Při vyšetření kinematiky dolní končetiny v hlubokém podřepu byla zaznamenána poloha segmentů v okamžiku, kdy kolenní kloub svírá pravý úhel ($90^\circ$).
* Délka stehenního segmentu (vzdálenost středu kyčelního a kolenního kloubu) je **$40\text{ cm}$**.
* Přímá vzdálenost mezi středem kyčelního kloubu a středem hlezenního kloubu (kotníku) činí **$50\text{ cm}$**.

Na základě uvedeného schématu určete velikost vektoru $\vec{u}$ představujícího funkční délku bérce.


```{figure} images/prx-016.png
:width: 350px
:align: center
:name: fig-sily-drep

Kinematický řetězec v dřepu.
```

````

````{solution} ex-sily-drep
:class: dropdown
**Řešení úlohy:**

Model kinematického řetězce dolní končetiny tvoří pravoúhlý trojúhelník s pravým úhlem v kolenním kloubu (*articulatio genus*):

* **Přepona ($c$):** přímá vzdálenost kyčel–kotník, $c = 50\text{ cm}$
* **Odvěsna 1 ($a$):** délka stehna (femuru), $a = 40\text{ cm}$
* **Odvěsna 2 ($b$):** délka bérce, $b = \vert{}\vec{u}\vert{}$

Podle Pythagorovy věty platí:

$$c^2 = a^2 + \vert{}\vec{u}\vert{}^2$$

Vyjádřením velikosti vektoru $\vert{}\vec{u}\vert{}$:

$$\vert{}\vec{u}\vert{} = \sqrt{c^2 - a^2}$$

Dosazení zadaných hodnot:

$$\vert{}\vec{u}\vert{} = \sqrt{50^2 - 40^2} = \sqrt{2500 - 1600} = \sqrt{900} = \mathbf{30\text{ cm}}$$


Velikost vektoru polohy bérce činí **$\vert{}\vec{u}\vert{} = 30\text{ cm}$** (odpovídá základnímu poměru stran pythagorejského trojúhelníku $3 : 4 : 5$).

````


````{exercise} Přetahování lanem
:label: ex-sily-lano

**Při přetahování lanem se sešla tři družstva (červené, žluté a zelené), která v uzlu působí těmito silami:**

* **Červené družstvo:** horizontální složka síly $F_{1x} = 1250\text{ N}$, vertikální složka síly $F_{1y} = 0\text{ N}$
* **Žluté družstvo:** horizontální složka síly $F_{2x} = -790\text{ N}$, vertikální složka síly $F_{2y} = -139\text{ N}$

**Úkol:**
Určete horizontální ($F_{3x}$) a vertikální ($F_{3y}$) složku síly třetího (zeleného) družstva tak, aby se střed lana nepohyboval (soustava zůstala ve statické rovnováze).


```{figure} images/prx-017.png
:width: 350px
:align: center
:name: fig-sily-lano

Kinematický řetězec v dřepu.
```

````

````{solution} ex-sily-lano
:class: dropdown

Aby se střed lana nepohyboval, musí být součet všech působících sil roven nule ($\sum \vec{F} = \vec{0}$).

**Horizontální směr:**


$$\sum F_x = F_{1x} + F_{2x} + F_{3x} = 0$$

$$1250\text{ N} + (-790\text{ N}) + F_{3x} = 0$$

$$460\text{ N} + F_{3x} = 0$$

$$F_{3x} = -460\text{ N}$$

**Vertikální směr:**


$$\sum F_y = F_{1y} + F_{2y} + F_{3y} = 0$$

$$0\text{ N} + (-139\text{ N}) + F_{3y} = 0$$

$$-139\text{ N} + F_{3y} = 0$$

$$F_{3y} = 139\text{ N}$$

Pro složky síly třetího družstva tedy musí platit:

* **Horizontální složka síly (hss):** **$-460\text{ N}$** (směřuje doleva)
* **Vertikální složka síly (vss):** **$139\text{ N}$** (směřuje nahoru)

````
