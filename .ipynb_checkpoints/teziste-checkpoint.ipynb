# Těžiště


````{exercise} Těžiště člověka s břemenem
:label: ex-teziste-bremeno

Člověk o hmotnosti $m_1 = 70\text{ kg}$ nese na hlavě břemeno o hmotnosti $m_2 = 30\text{ kg}$. Vzdálenost těžiště těla a těžiště břemena je $d = 1\text{ m}$. V jaké vzdálenosti od těžiště těla se nachází společné těžiště soustavy?

```{figure} images/pr-009.png
:width: 500px
:align: center
:name: fig-teziste-bremeno

Schéma rozložení hmotností člověka a břemene.
```

````


```{solution} ex-teziste-bremeno
:class: dropdown

Zvolme počátek v těžišti těla ($x_1 = 0$, $x_2 = 1\text{ m}$):

$$x_T = \frac{m_1 x_1 + m_2 x_2}{m_1 + m_2} = \frac{70 \cdot 0 + 30 \cdot 1{,}0}{70 + 30} = \frac{30}{100} = 0{,}30\text{ m} = 30\text{ cm}$$


```



````{exercise} Těžiště uholníku
:label: ex-uholniku
Na obrázku je pravítko (uholník) vyrobené z homogenního materiálu. Určete horizontální pozici těžiště od levého
dolního rohu pravítka.

```{figure} images/prx-004.png
:width: 250px
:align: center
:name: fig-teziste-uholnik

Uholník
```

````


```{solution} ex-teziste-bremeno
:class: dropdown

Pro určení vodorovné souřadnice těžiště $x_T$ vzhledem k levému dolnímu rohu (počátek $[0, 0]$) rozdělíme těleso tvaru písmene L na dva obdélníky.

**Rozdělení na části (svislý a vodorovný pás):**

* **Část 1 (svislé rameno po celé výšce):**
* Šířka: $4{,}0\text{ cm}$
* Výška: $18{,}0\text{ cm}$
* Plocha: $S_1 = 4{,}0 \cdot 18{,}0 = 72\text{ cm}^2$
* Vodorovná souřadnice těžiště: $x_1 = \frac{4{,}0}{2} = 2{,}0\text{ cm}$


* **Část 2 (zbývající vodorovné rameno vpravo):**
* Šířka: $12{,}0 - 4{,}0 = 8{,}0\text{ cm}$
* Výška: $4{,}0\text{ cm}$
* Plocha: $S_2 = 8{,}0 \cdot 4{,}0 = 32\text{ cm}^2$
* Vodorovná souřadnice těžiště: $x_2 = 4{,}0 + \frac{8{,}0}{2} = 8{,}0\text{ cm}$

**Výpočet polohy těžiště:**

$$x_T = \frac{S_1 \cdot x_1 + S_2 \cdot x_2}{S_1 + S_2}$$

$$x_T = \frac{72 \cdot 2{,}0 + 32 \cdot 8{,}0}{72 + 32} = \frac{144 + 256}{104} = \frac{400}{104} \approx 3{,}85\text{ cm}$$

Horizontální pozice těžiště od levého dolního rohu je **$\frac{50}{13}\text{ cm} \approx 3{,}85\text{ cm}$**.

* Proč počítáme ve vztahu pro těžiště s plochou a ne s hmotností? (Nápověda: Vyjádřete hmotnost pomocí plochy, tloušťky pravítka a hustoty materiálu.)


```
