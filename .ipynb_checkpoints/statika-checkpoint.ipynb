# Statická rovnováha

````{exercise} Zatížení meziobratlového disku
:label: ex-meziobratlovy-disk

Určete poměr $\frac{F_k}{F_s}$ velikostí kompresní síly $F_k$ a smykové síly $F_s$ působící v meziobratlovém disku, když víte, že velikost síly svalu *m. erector spinae* potřebná k udržení břemene $F_B = 100\text{ N}$ ve vodorovné poloze je $F_m = 1725\text{ N}$. Hmotnost hlavy je $m_H = 5{,}8\text{ kg}$, hmotnost trupu $m_T = 32{,}8\text{ kg}$ a hmotnost obou paží je $m_R = 8{,}1\text{ kg}$. Úhel $\alpha = 20^\circ$.

```{figure} images/pr-005.png
:width: 400px
:align: center
:name: fig-meziobratlovy-disk

Biomechanický model zatížení bederního segmentu.
```

````


```{solution} ex-meziobratlovy-disk
:class: dropdown

Předpokládáme statickou rovnováhu a to, že vektor síly zádového svalstva ($F_m$) působí rovnoběžně s kompresní silou ($F_k$), tedy kolmo na plochu meziobratlového disku L5/S1. Tíhové zrychlení uvažujeme $g = 9{,}81\text{ m/s}^2$.

1. Výpočet celkové vertikální zátěže

Nejprve určíme celkovou sílu $F_G$, která působí na horní polovinu těla svisle dolů. Ta je dána součtem tíhových sil hlavy, trupu, paží a samotného břemene.

* Celková hmotnost horní části těla: $m = m_H + m_T + m_R = 5{,}8 + 32{,}8 + 8{,}1 = 46{,}7\text{ kg}$
* Tíha horní části těla: $G = m \cdot g = 46{,}7 \cdot 9{,}81 \doteq 458{,}13\text{ N}$
* Celková vertikální zátěž: $F_G = G + F_B = 458{,}13 + 100 = 558{,}13\text{ N}$

2. Sestavení rovnic rovnováhy

Celkovou vertikální sílu $F_G$ rozložíme do souřadného systému natočeného o úhel $\alpha = 20^\circ$ (podle sklonu disku L5/S1):

* Složka působící **kompresně** (kolmo na disk): $F_{G,k} = F_G \cdot \cos\alpha$
* Složka působící **smykově** (rovnoběžně s diskem): $F_{G,s} = F_G \cdot \sin\alpha$

Z podmínek rovnováhy sil pro ploténku L5/S1 vyplývá:

* **Kompresní síla ($F_k$):** Reakce disku musí vyrovnat celkový tlak, který je tvořen normálovou složkou vertikální zátěže a tahem zádového svalu.

$$F_k = F_m + F_G \cdot \cos\alpha$$


* **Smyková síla ($F_s$):** Reakce disku musí zabránit posunu obratlů, tedy vyrovnat tečnou (smykovou) složku vertikální zátěže.

$$F_s = F_G \cdot \sin\alpha$$

3. Číselný výpočet

Dosadíme známé hodnoty do rovnic:

* Výpočet kompresní síly:

$$F_k = 1725 + 558{,}13 \cdot \cos(20^\circ)$$


$$F_k = 1725 + (558{,}13 \cdot 0{,}9397)$$


$$F_k \doteq 1725 + 524{,}47 = 2249{,}47\text{ N}$$


* Výpočet smykové síly:

$$F_s = 558{,}13 \cdot \sin(20^\circ)$$


$$F_s = 558{,}13 \cdot 0{,}3420 \doteq 190{,}88\text{ N}$$

4. Výsledný poměr sil

Nakonec určíme požadovaný poměr velikostí těchto dvou sil:

$$\frac{F_k}{F_s} = \frac{2249{,}47}{190{,}88} \doteq 11{,}78$$

Poměr kompresní síly ke smykové síle působící v meziobratlovém disku L5/S1 je přibližně **11,8**.


Dávalo by smysl tento zjednodušený model rozšířit například o vliv nitrobřišního tlaku? Proč nemusíme počítat momentovou rovnováhu?

```



````{exercise} Rovnováha veslařského pádla
:label: ex-rovnovaha-padla

Síly, které působí na pádlo, jsou znázorněny na obrázku. Určete zatížení závěsu $F_0$, je-li síla vyvíjená veslařem $F_H = 100\text{ N}$, $L_1 = 30\text{ cm}$, $L_2 = 1\text{ m}$ a pádlo je v rovnováze. Vysvětlete co je špatně na obrázku. 

```{figure} images/pr-010.jpg
:width: 500px
:align: center
:name: fig-rovnovaha-padla

Pákový mechanismus veslařského pádla.
```

````


```{solution} ex-rovnovaha-padla
:class: dropdown

Pádlo (veslo) můžeme modelovat jako dvojzvratnou páku a pro výpočet využijeme základní podmínky statické rovnováhy.

1. Výpočet síly odporu vody ($F_B$)

Nejprve použijeme **momentovou větu**. Osu otáčení si zvolíme v místě závěsu (v působišti síly $F_0$). Aby se pádlo neotáčelo, musí být moment síly, kterou působí veslař, v rovnováze s momentem síly odporu vody:

$$F_H \cdot L_1 = F_B \cdot L_2$$

Z rovnice si vyjádříme neznámou sílu $F_B$ (před dosazením je nutné sjednotit jednotky délky, např. na metry: $L_1 = 0{,}3\text{ m}$, $L_2 = 1\text{ m}$):

$$F_B = F_H \cdot \frac{L_1}{L_2}$$

$$F_B = 100 \cdot \frac{0{,}3}{1} = 30\text{ N}$$

Odpor vody na listu pádla je tedy $30\text{ N}$.


2. Výpočet zatížení závěsu ($F_0$)

Nyní aplikujeme **podmínku silové rovnováhy**. Aby se pádlo neposouvalo, musí být součet sil působících jedním směrem roven součtu sil působících směrem opačným. Z nákresu je zřejmé, že síly $F_H$ a $F_B$ působí stejným směrem (dolů), zatímco reakce v závěsu $F_0$ působí proti nim (nahoru):

$$F_0 = F_H + F_B$$

$$F_0 = 100 + 30 = 130\text{ N}$$

**Výsledek:**
Celkové zatížení závěsu $F_0$ je **130 N**.

```


````{exercise} Síla v hamstringách
:label: ex-sila-hamstringy

Určete sílu v ischiokrurálních svalech (hamstringy) $F_m$ při zatížení podle obrázku, je-li tíha břemene $W = 100\text{ N}$ a rozměry odpovídají schématu.

```{figure} images/pr-015.png
:width: 250px
:align: center
:name: fig-sila-hamstringy

Pákový model dolní končetiny a hamstringů.
```

````


```{solution} ex-sila-hamstringy
:class: dropdown

Pro výpočet svalové síly $F_m$ vyjdeme z momentové podmínky rovnováhy k ose rotace v kolenním kloubu.

**Zadané hodnoty:**

* Tíhová síla břemene: $W = 100\text{ N}$
* Rameno síly břemene: $r_W = 40\text{ cm} = 0{,}40\text{ m}$
* Rameno síly hamstringů: $r_m = 5\text{ cm} = 0{,}05\text{ m}$

**Momentová rovnováha:**
Vzhledem k ose otáčení v koleni musí být součet statických momentů roven nule:


$$\sum M_O = 0$$

$$F_m \cdot r_m - W \cdot r_W = 0$$

$$F_m = \frac{W \cdot r_W}{r_m}$$

Po dosazení:


$$F_m = \frac{100 \cdot 0{,}40}{0{,}05} = \frac{40}{0{,}05} = 800\text{ N}$$

Tahová síla v hamstringu činí **800 N**.

```

````{exercise} Síla ve vzpřimovačech páteře
:label: ex-sila-vzprimovace

Člověk v předkloně zvedá břemeno o tíze 200 N. Záda jsou vodorovně. Předpokládejme, že musculus
erector spinae se upíná v ⅔ délky páteře a zvírá s ní úhel 12°. Tíha horné části těla je 350 N a těžište je v
polovině délky páteře. Určete sílu v musculus erector spinae.

```{figure} images/prx-005.png
:width: 250px
:align: center
:name: fig-sila-hamstringy

Zatížení páteře v předklonu.
```

````


```{solution} ex-sila-vzprimovace
:class: dropdown

**Krok 1: Analýza obrázku a zadání**

Obrázek (Panel B) nám poskytuje přesné schéma sil a jejich ramen vzhledem k ose **O** (kyčelní kloub).

* **Pánev/Kyčle (bod O):** Osa otáčení. Zde působí reakční síly $R_x$ a $R_y$, které ale nemají moment vzhledem k bodu O (jejich rameno je nula).
* **Páteř:** V modelu je reprezentována vodorovným nosníkem o celkové délce, kterou označíme **$L$**.
* **Tíha břemene ($W_{load}$):** Působí na konci páteře (ruce). Směřuje svisle dolů. Hodnota: $W_{load} = 200 \text{ N}$. Rameno této síly je celá délka páteře, **$L$**. Na obrázku je to vzdálenost od O k modré šipce 200 N.
* **Tíha horní části těla ($W_{body}$):** Působí v polovině délky páteře. Směřuje svisle dolů. Hodnota: $W_{body} = 350 \text{ N}$. Rameno této síly je **$\frac{L}{2}$**. Na obrázku je to vzdálenost od O k modré šipce 350 N.
* **Síla svalu ($T$ nebo $F_{m}$):** Toto je naše neznámá síla v *musculus erector spinae*. Působí v $\frac{2}{3}$ délky páteře. Směřuje šikmo vzhůru pod úhlem $12^\circ$ k vodorovné ose (páteři). Rameno působiště svalu je **$\frac{2L}{3}$**. Na obrázku je to vzdálenost od O k patě černé šipky $T$.

**Krok 2: Stanovení momentové věty**

Zvolíme směr otáčení: Moment sil táhnoucích dolů (tíhy) budeme považovat záporný, protože vůči bodu O otáčí ve směru hodinových ručiček zatímco moment svalu táhnoucího nahoru za kladný. V rovnováze se musí rovnat.

Moment síly = Síla $\times$ Rameno síly $\times$ sinus úhlu mezi silou a ramenem.

1. **Moment tíhy těla ($M_{body}$):** Působí kolmo na vodorovnou páteř, úhel je $90^\circ$, $\sin(90^\circ)=1$.

$$M_{body} = - W_{body} \cdot \left(\frac{L}{2}\right) =  - 350 \cdot \frac{L}{2}$$


2. **Moment tíhy břemene ($M_{load}$):** Působí kolmo na vodorovnou páteř, úhel je $90^\circ$.

$$M_{load} = -W_{load} \cdot L = - 200 \cdot L$$


3. **Moment síly svalu ($M_{muscle}$):** Svalu působí pod úhlem $12^\circ$, musíme vzít *svislou složku* této síly, která skutečně vytváří moment proti rotaci dolů. Tato složka je $T \cdot \sin(12^\circ)$.

$$M_{muscle} = + \left(T \cdot \sin(12^\circ)\right) \cdot \left(\frac{2L}{3}\right)$$



**Krok 3: Výpočet**

Podle momentové věty: Součet kladných momentů + Součet záporných momentů = 0.


$$M_{body} + M_{load} + M_{muscle} = 0$$

Dosadíme vztahy z Kroku 2:


$$ -350 \cdot \frac{L}{2} - 200 \cdot L + \left(T \cdot \sin(12^\circ)\right) \cdot \left(\frac{2L}{3}\right)= 0$$

Všimněte si, že **neznámá délka páteře $L$ se vyskytuje ve všech členech rovnice a můžeme ji vykrátit**. To je klíčový krok:

$$350 \cdot \frac{1}{2} + 200 = \frac{2}{3} \cdot T \cdot \sin(12^\circ)$$

Nyní vypočítáme levou stranu:


$$\frac{350}{2} + 200 = 175 + 200 = 375 \text{ N}$$

Rovnice nyní vypadá takto:


$$375 = \frac{2}{3} \cdot T \cdot \sin(12^\circ)$$

Z této rovnice izolujeme neznámou sílu $T$:


$$T = \frac{375 \cdot 3}{2 \cdot \sin(12^\circ)}$$

Použijeme kalkulačku pro výpočet $\sin(12^\circ) \approx 0{,}2079$:

$$T = \frac{1125}{2 \cdot 0{,}2079} = \frac{1125}{0{,}4158} \approx 2705{,}6 \text{ N}$$

**Výsledek**

Síla v *musculus erector spinae* při zvedání tohoto břemene je přibližně **$2706 \text{ N}$**.

*To odpovídá tahu, který by vyvolalo závaží o hmotnosti přibližně $270 \text{ kg}$. To ukazuje, jak obrovským silám musí zádové svaly čelit kvůli velmi nevýhodnému (ostrému) úhlu jejich upnutí vzhledem k páteři při hlubokém předklonu.*

```

````{exercise} Síla v podpěře
:label: ex-sila-fakir

Člověk o celkové výšce $L = 2{,}0\text{ m}$ a celkové tíze $G = 1000\text{ N}$ leží v horizontální poloze podepřený pouze ve dvou místech – pod zátylkem (podpora 1) a pod patami (podpora 2). Pro zjednodušenou biomechanickou analýzu je tělo rozděleno do čtyř hlavních segmentů, jejichž tíhové síly působí v příslušných těžištích podle obrázku. Vypočítejte velikost reakční síly **$F_1$** působící v podpoře pod hlavou.

```{figure} images/prx-008.png
:width: 550px
:align: center
:name: fig-sila-hamstringy

Biomechanický model ležícího těla s rozložením sil
```

````


```{solution} ex-sila-fakir
:class: dropdown

**Řešení statického modelu ležícího těla**

---

**1. Souřadnice působišť sil od levé podpory ($F_1$ v bodě $x = 0{,}0\text{ m}$):**

* Podpora 1: $x_1 = 0{,}0\text{ m}$
* Hlava a hrudník: $x_A = 0{,}35\text{ m}$
* Břicho a bedra: $x_B = 0{,}35 + 0{,}45 = 0{,}80\text{ m}$
* Pánev a stehna: $x_C = 0{,}80 + 0{,}50 = 1{,}30\text{ m}$
* Bérce a chodidla: $x_D = 1{,}30 + 0{,}45 = 1{,}75\text{ m}$
* Podpora 2: $x_2 = 1{,}75 + 0{,}25 = 2{,}00\text{ m}$

---

**2. Určení reakční síly $F_2$ z momentové podmínky k bodu podpory 1 ($x_1 = 0$):**

Součet momentů všech sil vzhledem k ose procházející bodem $1$ musí být roven nule ($\sum M_1 = 0$):

$$F_2 \cdot x_2 - (F_A \cdot x_A + F_B \cdot x_B + F_C \cdot x_C + F_D \cdot x_D) = 0$$

$$F_2 \cdot 2{,}0 = 320 \cdot 0{,}35 + 200 \cdot 0{,}80 + 300 \cdot 1{,}30 + 180 \cdot 1{,}75$$

Vyčíslení jednotlivých momentů tíhových sil:

* $M_A = 320 \cdot 0{,}35 = 112{,}0\text{ N}\cdot\text{m}$
* $M_B = 200 \cdot 0{,}80 = 160{,}0\text{ N}\cdot\text{m}$
* $M_C = 300 \cdot 1{,}30 = 390{,}0\text{ N}\cdot\text{m}$
* $M_D = 180 \cdot 1{,}75 = 315{,}0\text{ N}\cdot\text{m}$

$$F_2 \cdot 2{,}0 = 112{,}0 + 160{,}0 + 390{,}0 + 315{,}0 = 977{,}0\text{ N}\cdot\text{m}$$

$$F_2 = \frac{977{,}0}{2{,}0} = \mathbf{488{,}5\text{ N}}$$

---

**3. Určení reakční síly $F_1$:**

**Z podmínky rovnováhy sil ve svislém směru ($\sum F_y = 0$):**


$$F_1 + F_2 - (F_A + F_B + F_C + F_D) = 0$$

$$F_1 + 488{,}5 - 1000 = 0$$

$$F_1 = 1000 - 488{,}5 = \mathbf{511{,}5\text{ N}}$$

*(Alternativně z momentové podmínky k bodu podpory 2 při $x_2 = 2{,}0\text{ m}$):*


$$F_1 \cdot 2{,}0 = F_A \cdot (2{,}0 - 0{,}35) + F_B \cdot (2{,}0 - 0{,}80) + F_C \cdot (2{,}0 - 1{,}30) + F_D \cdot (2{,}0 - 1{,}75)$$

$$F_1 \cdot 2{,}0 = 320 \cdot 1{,}65 + 200 \cdot 1{,}20 + 300 \cdot 0{,}70 + 180 \cdot 0{,}25$$

$$F_1 \cdot 2{,}0 = 528 + 240 + 210 + 45 = 1023\text{ N}\cdot\text{m}$$

$$F_1 = \frac{1023}{2{,}0} = \mathbf{511{,}5\text{ N}}$$

---

**4. Výsledná poloha celkového těžiště těla ($x_T$):**

$$x_T = \frac{\sum (F_i \cdot x_i)}{\sum F_i} = \frac{977{,}0\text{ N}\cdot\text{m}}{1000\text{ N}} = \mathbf{0{,}977\text{ m}}$$

Těžiště leží ve vzdálenosti **$0{,}977\text{ m}$** od temene hlavy (přibližně $48{,}8\,\%$ celkové tělesné výšky).

```

````{exercise} Určení síly v deltovém svalu a zatížení glenohumerálního kloubu
:label: ex-sila-delt


Určete tahovou sílu v deltovém svalu (*musculus deltoideus*) a celkovou reakční sílu v ramenním kloubu včetně jejího směru.

* Hmotnost horní končetiny: $m = 3{,}6\text{ kg}$
* Vzdálenost těžiště končetiny od ramenního kloubu: $L_1 = 0{,}66\text{ m}$
* Vzdálenost úponu deltového svalu na humeru: $L_2 = 0{,}33\text{ m}$
* Úhel úponu deltového svalu vůči podélné ose humeru: $\theta = 20^\circ$
* Úhel odklonu humeru od svislice těla: $\alpha = 80^\circ$
* Tíhové zrychlení: $g = 9{,}81\text{ m}\cdot\text{s}^{-2}$

```{figure} images/prx-015.svg
:width: 350px
:align: center
:name: fig-sila-delt

Měření síly v bicepsu.
```

````


```{solution} ex-sila-delt
:class: dropdown

Tíhová síla celé paže působící v těžišti:

$$W = m \cdot g = 3{,}6 \cdot 9{,}81 \doteq 35{,}32\text{ N}$$


**1. Výpočet síly v deltovém svalu ($F_D$):**

Zvolíme ramenní kloub $O$ jako střed otáčení. Pro statickou rovnováhu momentů sil platí:

$$\sum M_O = 0 \implies F_D \cdot \sin(\theta) \cdot L_2 - W \cdot \sin(\alpha) \cdot L_1 = 0$$

kde:

* $L_1 \cdot \sin(\alpha)$ je vodorovné rameno tíhové síly $W$,
* $F_D \cdot \sin(\theta)$ je složka svalové síly kolmá na kost pažní.

Vyjádřením svalové síly $F_D$:

$$F_D = W \cdot \frac{L_1 \cdot \sin(\alpha)}{L_2 \cdot \sin(\theta)}$$

$$F_D = 35{,}32 \cdot \frac{0{,}66 \cdot \sin(80^\circ)}{0{,}33 \cdot \sin(20^\circ)} = 35{,}32 \cdot \frac{0{,}66 \cdot 0{,}9848}{0{,}33 \cdot 0{,}3420} \doteq \mathbf{203{,}4\text{ N}}$$


**2. Výpočet celkové síly v ramenním kloubu ($\vec{R}$):**

Zvolíme souřadnicový systém s počátkem v kloubu $O$:

* Osa $x$ směřuje vodorovně ven od trupu (doprava).
* Osa $y$ směřuje svisle nahoru.

Směrový úhel svalu $\vec{F}_D$ vůči svislici těla je $\alpha - \theta = 80^\circ - 20^\circ = 60^\circ$. Síla svalu působí směrem k trupu a nahoru:

* $F_{Dx} = -F_D \cdot \sin(60^\circ) = -203{,}38 \cdot 0{,}8660 \doteq -176{,}13\text{ N}$
* $F_{Dy} = +F_D \cdot \cos(60^\circ) = +203{,}38 \cdot 0{,}5000 \doteq +101{,}69\text{ N}$

Tíhová síla $\vec{W}$ působí svisle dolů:

* $W_x = 0\text{ N}$
* $W_y = -35{,}32\text{ N}$

Z podmínek rovnováhy sil pro ramenní kloub ($\vec{R} + \vec{F}_D + \vec{W} = \vec{0}$):

* **Vodorovná složka reakce:**

$$R_x = -F_{Dx} = +176{,}13\text{ N}$$


* **Svislá složka reakce:**

$$R_y = -F_{Dy} - W_y = -101{,}69 - (-35{,}32) = -66{,}37\text{ N}$$


**Velikost výsledné síly v kloubu:**


$$R = \sqrt{R_x^2 + R_y^2} = \sqrt{176{,}13^2 + (-66{,}37)^2} = \sqrt{31021{,}8 + 4405{,}0} \doteq \mathbf{188{,}2\text{ N}}$$

**Směr kloubní síly:**

* Odchylka vektoru $\vec{R}$ pod vodorovnou rovinu:

$$\varphi = \operatorname{arctg}\left(\frac{\vert{}R_y\vert{}}{\vert{}R_x\vert{}}\right) = \operatorname{arctg}\left(\frac{66{,}37}{176{,}13}\right) \doteq \mathbf{20{,}6^\circ}$$



*(nebo $69{,}4^\circ$ od svislice směrem dolů a laterálně do kloubní jamky).*

**Shrnutí výsledků:**

* Tahová síla v deltovém svalu: **$F_D \doteq 203\text{ N}$**
* Celková kontaktní síla v ramenním kloubu: **$R \doteq 188\text{ N}$**
* Směr síly v kloubu: působí do kloubního pouzdra pod úhlem **$20{,}6^\circ$ dolů od vodorovné roviny**

```
