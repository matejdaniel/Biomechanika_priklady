# Moment síly


````{exercise} Utahovací moment matice
:label: ex-utahovani-matice

Při utahování matice působí síly podle obrázku. Určete výsledný moment sil působící na matici.

```{figure} images/pr-012.jpg
:width: 800px
:align: center
:name: fig-utahovani-matice

Utahování matice silovou dvojicí.
```

````


```{solution} ex-utahovani-matice
:class: dropdown

**Zadané hodnoty z obrázku:**

* Působiště osy otáčení: střed matice (zcela vlevo).
* Síla v místě A: $F_A = 50\text{ N}$ (směrem dolů).
* Rameno síly v místě A (vzdálenost od osy): $d_A = 20\text{ cm} = 0{,}2\text{ m}$.
* Síla v místě B: $F_B = 60\text{ N}$ (směrem dolů).
* Rameno síly v místě B (vzdálenost od osy): $d_B = 30\text{ cm} = 0{,}3\text{ m}$.

### Řešení

1. **Princip:** Moment síly $M$ se vypočítá jako součin velikosti síly $F$ a ramene síly $d$ (kolmé vzdálenosti od osy otáčení). Výsledný moment je dán součtem momentů jednotlivých sil. Protože obě síly působí stejným směrem (dolů) a snaží se matici otáčet stejným směrem (po směru hodinových ručiček), jejich účinky se sčítají.

2. **Vzorec:**

$$M_{\text{výsl}} = M_A + M_B$$


$$M_{\text{výsl}} = (F_A \cdot d_A) + (F_B \cdot d_B)$$


3. **Dosazení:**

Je důležité dosadit vzdálenosti v základních jednotkách, tedy v metrech.

$$M_{\text{výsl}} = (50\text{ N} \cdot 0{,}2\text{ m}) + (60\text{ N} \cdot 0{,}3\text{ m})$$


$$M_{\text{výsl}} = 10\text{ Nm} + 18\text{ Nm}$$


$$M_{\text{výsl}} = 28\text{ Nm}$$

### Odpověď

Výsledný moment sil působící na matici je **28 Nm**.

```


````{exercise} Moment působící na pálku
:label: ex-moment-palka

Hráč drží v ruce pálku s tíhou 10 N. Těžiště pálky se nachází ve vzdálenosti 60 cm od bodu O. Určete
celkový moment síly, kterým působí hráč na pálku.Pálka je v rovnováze.

```{figure} images/prx-002.png
:width: 250px
:align: center
:name: fig-sila-hamstringy

Pálka.
```

````


```{solution} ex-moment-palka
:class: dropdown

Aby byla pálka v klidové rovnováze, musí být celkový výsledný moment všech působících sil vzhledem k libovolnému bodu (zde bodu $O$) roven nule:

$$\sum M_O = 0$$

**Zadané hodnoty:**

* Tíhová síla pálky: $W = 10\text{ N}$
* Vzdálenost těžiště (rameno tíhové síly): $d = 60\text{ cm} = 0{,}6\text{ m}$

**Moment tíhové síly:**
Tíhová síla působí směrem dolů a způsobuje moment (otáčení po směru hodinových ručiček):


$$M_W = W \cdot d = 10\text{ N} \cdot 0{,}6\text{ m} = 6\text{ N}\cdot\text{m}$$

**Moment působený hráčem:**
Aby nastala rovnováha, musí hráč v úchopu (kolem bodu $O$) působit opačným momentem stejné velikosti:


$$M_{\text{hráč}} - M_W = 0 \implies M_{\text{hráč}} = M_W = 6\text{ N}\cdot\text{m}$$

Celkový moment síly, kterým hráč na pálku působí, má velikost **$6\text{ N}\cdot\text{m}$** (směřuje proti směru otáčení způsobeného tíhou pálky, tedy proti směru hodinových ručiček).

```

````{exercise} Moment na hlavici femuru
:label: ex-moment-hlavice

Na hlavici femuru působí síla 2 kN pod úhlem 60° k podélné ose krčku femuru. Určete moment síly ve vzdálenosti 3 cm od působiště síly. 

```{figure} images/prx-011.png
:width: 350px
:align: center
:name: fig-moment-hlavice

Síly působící na femur při trakci.
```

````

````{solution} ex-moment-hlavice
:class: dropdown

**Zadané hodnoty:**

* Kontaktní síla v kyčelním kloubu: $F = 2000\text{ N}$
* Úhel síly vůči podélné ose krčku femuru: $\alpha = 60^\circ$
* Vzdálenost bodu od působiště síly podél osy krčku: $d = 3\text{ cm} = 0{,}03\text{ m}$

---

**Fyzikální rozbor:**

Sílu $\vec{F}$ rozložíme do dvou složek vzhledem k podélné ose krčku femuru:

* **Normálová (osová/tlaková) složka:**
$$F_\parallel = F \cdot \cos(\alpha)$$


Působí podél osy krčku, prochází uvažovaným bodem a nevyvolává ohybový moment.
* **Příčná (tangenciální/ohybová) složka:**
$$F_\perp = F \cdot \sin(\alpha)$$


Působí kolmo na osu krčku a vyvolává ohybový moment na rameni $d$.

---

**Výpočet momentu síly:**

Velikost momentu síly k danému bodu je dána vztahem:

$$M = F_\perp \cdot d = F \cdot d \cdot \sin(\alpha)$$

Dosazení číselných hodnot:

$$M = 2000 \cdot 0{,}03 \cdot \sin(60^\circ)$$

$$M = 60 \cdot \frac{\sqrt{3}}{2} = 30\sqrt{3} \doteq 51{,}96\text{ N}\cdot\text{m}$$

---

Moment síly působící v daném průřezu krčku femuru je přibližně **$52\text{ N}\cdot\text{m}$** (respektive $51{,}96\text{ N}\cdot\text{m}$).

````


````{exercise} Rotační složka síly
:label: ex-moment-rot

Kterou složkou síly nejlépe popíšeme rotační účinek bicepsu vůči loketnímu kloubu?

```{figure} images/prx-019.png
:width: 350px
:align: center
:name: fig-moment-rot

Možné složky svalové síly. 

```

````

````{solution} ex-moment-rot
:class: dropdown

Největší rotační účinek (moment síly k ose loketního kloubu) má složka **$F_B$** (svislá síla směřující nahoru).

* **Rameno síly a úhel:** Předloktí je ve vodorovné poloze. Rotační účinek je dán složkou síly, která působí kolmo na podélnou osu předloktí, tedy přesně ve svislém směru nahoru ($F_B$). Její rameno k ose otáčení v lokti je maximální možné ($r$), takže generuje moment $M = F_B \cdot r$.
* **Srovnání s ostatními složkami:**
* **$F_A$ (vodorovná):** Působí podél osy kosti předloktí; rameno síly k ose kloubu je nulové, takže vyvolává pouze kompresi či translaci v kloubu, nikoli rotaci ($M = 0$).
* **$F_C$ (ve směru svalu):** Působí šikmo pod úhlem menším než 90° vůči kosti, takže část síly se ztrácí v osovém tahu a její kolmá (rotační) komponenta je menší než celá velikost síly.
* **$F_D$ (kolmo na biceps):** Směřuje šikmo a netvoří pravý úhel s vodorovnou kostí předloktí, její efektivní kolmé rameno vůči páce předloktí je proto menší než u ryze svislého vektoru $F_B$.
````
