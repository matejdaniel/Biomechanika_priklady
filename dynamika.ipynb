````{exercise} Pohyb s třením
:label: ex-sila-treni

Sportovec tlačí po vodorovném povrchu tělocvičny tréninkový biomechanický trenažér (sáňky se zátěží) o celkové hmotnosti $m = 9{,}0\text{ kg}$. Působí na něj vodorovnou silou $F = 25\text{ N}$ a uděluje mu zrychlení $a = 1{,}5\text{ m}\cdot\text{s}^{-2}$. Vypočítejte koeficient dynamického smykového tření mezi kontaktní plochou trenažéru a podlahou.

```{figure} images/prx-009.png
:width: 550px
:align: center
:name: fig-sila-treni

Sportovec tlačí po vodorovném povrchu tělocvičny tréninkový biomechanický trenažér.  
```

````


```{solution} ex-sila-treni
:class: dropdown

**Zadání:**

* Hmotnost trenažéru se zátěží: $m = 90\text{ kg}$
* Působící tlačná síla: $F = 525\text{ N}$
* Zrychlení trenažéru: $a = 1{,}5\text{ m}\cdot\text{s}^{-2}$
* Tíhové zrychlení: $g = 9{,}81\text{ m}\cdot\text{s}^{-2}$ (popř. $10\text{ m}\cdot\text{s}^{-2}$)
* Hledaná veličina: koeficient dynamického smykového tření $f$



**1. Pohybová rovnice ve směru pohybu (vodorovná osa):**

Na trenažér působí ve směru pohybu tlačná síla sportovce $F$ a proti směru pohybu třecí síla $F_t$:

$$F - F_t = m \cdot a$$

Z této rovnice vyjádříme velikost třecí síly:

$$F_t = F - m \cdot a$$

Dosazení číselných hodnot:


$$F_t = 525 - 90 \cdot 1{,}5 = 525 - 135 = 390\text{ N}$$



**2. Rovnováha ve svislém směru:**

Protože se trenažér ve svislém směru nepohybuje, normálová tlaková reakce podložky $F_n$ je v rovnováze s tíhovou silou $F_g$:

$$F_n = F_g = m \cdot g$$

Při použití $g = 9{,}81\text{ m}\cdot\text{s}^{-2}$:


$$F_n = 90 \cdot 9{,}81 = 882{,}9\text{ N}$$

*(Při zaokrouhlení na $g = 10\text{ m}\cdot\text{s}^{-2}$: $F_n = 90 \cdot 10 = 900\text{ N}$)*



**3. Výpočet koeficientu dynamického tření $f$:**

Vztah pro smykové tření:


$$F_t = f \cdot F_n \implies f = \frac{F_t}{F_n} = \frac{F - m \cdot a}{m \cdot g}$$

* **Přesný výpočet ($g = 9{,}81\text{ m}\cdot\text{s}^{-2}$):**

$$f = \frac{390}{882{,}9} \approx \mathbf{0{,}442}$$


* **Výpočet se zaokrouhleným $g = 10\text{ m}\cdot\text{s}^{-2}$:**

$$f = \frac{390}{900} = \frac{13}{30} \approx \mathbf{0{,}433}$$





Koeficient dynamického smykového tření mezi trenažérem a podlahou činí přibližně **0,44** (resp. **0,43** při $g = 10\text{ m}\cdot\text{s}^{-2}$).

```

````{exercise} Hmotnost těla při volném pádu
:label: ex-hmotnost-pad

Na obrázku je znázorněna závislost uražené dráhy $s$ na čase $t$ při volném pádu sportovce. Určete jeho hmotnost když zanedbáme tření.
    
```{figure} images/prx-013.png
:width: 550px
:align: center
:name: fig-hmotnost-pad

Graf závislosti dráhy na čase volného pádu
```

````


```{solution} ex-hmotnost-pad
:class: dropdown

**Řešení:**

Hmotnost objektu **nelze z grafu určit**.

1. **Pohybová rovnice:**
Při volném pádu v homogenním tíhovém poli (se zanedbáním odporu prostředí) působí na těleso pouze tíhová síla:

$$F_G = m \cdot g$$



Podle 2. Newtonova zákona platí:

$$m \cdot a = m \cdot g \implies a = g$$



Zrychlení $a$ padajícího tělesa je rovno tíhovému zrychlení $g \doteq 9{,}81\text{ m}\cdot\text{s}^{-2}$ a hmotnost $m$ se vykrátí.

2. **Závislost dráhy na čase:**
Pro dráhu rovnoměrně zrychleného přímočarého pohybu z klidu ($v_0 = 0\text{ m}\cdot\text{s}^{-1}$) platí vztah:

$$s(t) = \frac{1}{2} g t^2$$



Křivka závislosti $s(t)$ je parabola jednoznačně definovaná pouze parametrem $g$. Je shodná pro těleso libovolné hmotnosti (např. pero ve vakuu i těžké závaží).

---

**Závěr:**

Závislost dráhy na čase volného pádu neobsahuje žádnou informaci o hmotnosti $m$, proto hmotnost z tohoto grafu určit nelze.
```

````{exercise} Moment setrvačnosti těla
:label: ex-moment-setrvacnosti

Ve které pozici má lidské tělo minimální moment setrvačnosti? Je naznačena osa rotace vzhledem ke
které je určován moment setrvačnosti.
    
```{figure} images/prx-014.png
:width: 550px
:align: center
:name: fig-moment-stervacnosti

Čtyři různé polohy lidského těla, (a), (b), (c) a (d), s vyznačenými hlavními osami rotace.
```
````


```{solution} ex-moment-setrvacnosti
:class: dropdown

Minimální moment setrvačnosti má lidské tělo v pozici **(d)**.

---

** Biomechanické důvodnění:**

Moment setrvačnosti tělesa je obecně definován vztahem:

$$I = \sum m_i r_i^2 \quad \left(\text{resp. } I = \int r^2 \, \mathrm{d}m\right)$$

kde $r_i$ představuje kolmou vzdálenost jednotlivých hmotnostních elementů $m_i$ od zvolené osy rotace. Závisí tedy na kvadrátu vzdálenosti hmoty od dané osy:

* **V pozici (d)** prochází osa rotace vertikálně (podélná/longitudinální osa těla). Vzpřímené tělo má naprostou většinu své hmotnosti (trup, hlava, končetiny) soustředěnou velmi blízko této ose (poloměr rotace $r$ je v řádu jednotek až desítek centimetrů). Moment setrvačnosti vůči této ose je u průměrného člověka nejmenší – typicky kolem **$1{,}0$ až $1{,}2\text{ kg}\cdot\text{m}^2$** (využívá se např. při rychlých piruetách v krasobruslení).
* **V pozicích (a) a (b)** prochází osa rotace příčně středem těla (transverzální osa). Tělo je natažené, takže hlava i dolní končetiny se nacházejí ve velké vzdálenosti od osy ($r$ dosahuje až cca $0{,}9\text{ m}$), což vede k maximálnímu momentu setrvačnosti (řádově **$10$ až $15\text{ kg}\cdot\text{m}^2$**).
* **V pozici (c)** je tělo zabalené do klubíčka (skrčka), což vzdálenost hmoty od transverzální osy výrazně redukuje oproti pozicím (a) a (b), moment setrvačnosti zde klesá na zhruba **$3$ až $4\text{ kg}\cdot\text{m}^2$**, stále je však výrazně vyšší než podél longitudinální osy v pozici (d).

```
