
```{solution} ex-deformace-kosti
:class: dropdown

Plocha průřezu kosti:
$$S = \pi r^2 = \pi \left(\frac{0{,}05}{2}\right)^2 \approx 1{,}963 \cdot 10^{-3}\text{ m}^2$$

Normálové napětí:
$$\sigma = \frac{F}{S} = \frac{100}{1{,}963 \cdot 10^{-3}} \approx 50930\text{ Pa}$$

Poměrné prodloužení:
$$\varepsilon = \frac{\sigma}{E} = \frac{50930}{10 \cdot 10^9} \approx 5{,}09 \cdot 10^{-6} \approx 5 \cdot 10^{-6}$$

```


````{exercise} Prodloužení Kirschnerova drátu
:label: ex-prodlouzeni-dratu

Ocelový drát má délku 1,12 m a má plochu průřezu 6,0 10$^{-3}$ cm$^2$. Youngův modul pružnosti je 20 1010
N/m$^2$. Pod tahem 115 N určete prodloužení drátu.

```{figure} images/prx-003.png
:width: 250px
:align: center
:name: fig-prodlouzeni-dratu

Experiment pro měření prodloužení drátu.
```

````

````{solution} ex-prodlouzeni-dratu
:class: dropdown

Prodloužení drátu $\Delta L$ určíme pomocí Hookeova zákona pro tah:

$$\Delta L = \frac{F \cdot L}{S \cdot E}$$

**Zadané hodnoty v základních jednotkách SI:**

* Počáteční délka: $L = 1{,}12\text{ m}$
* Průřez: $S = 6{,}0 \cdot 10^{-3}\text{ cm}^2 = 6{,}0 \cdot 10^{-3} \cdot 10^{-4}\text{ m}^2 = 6{,}0 \cdot 10^{-7}\text{ m}^2$
* Youngův modul pružnosti: $E = 20 \cdot 10^{10}\text{ N/m}^2 = 2{,}0 \cdot 10^{11}\text{ Pa}$
* Tahová síla: $F = 115\text{ N}$

**Výpočet:**


$$\Delta L = \frac{115 \cdot 1{,}12}{6{,}0 \cdot 10^{-7} \cdot 20 \cdot 10^{10}}$$

$$\Delta L = \frac{128{,}8}{1{,}2 \cdot 10^5} \approx 1{,}073 \cdot 10^{-3}\text{ m}$$

Prodloužení drátu je přibližně **$1{,}07\text{ mm}$** ($1{,}07 \cdot 10^{-3}\text{ m}$).

````
