# Energie



````{exercise} Veletoč na hrazdě
:label: ex-veletoc

Gymnasta provádí na hrazdě prvek veletoč. Jak možné provést daný prvek, když dochází ke ztrátě energie třením

```{figure} images/pr-001.png
:width: 250px
:align: center
:name: fig-veletoc

Schéma fázového rozpadu veletoče na hrazdě. Převzato z [Burton a kol, 2023](https://doi.org/10.1080/02640414.2023.2259201).
````

````{solution} ex-veletoc
:class: dropdown

Aby gymnasta mohl provést veletoč (giant swing) a překonat ztráty mechanické energie způsobené třením dlaní o hrazdu a aerodynamickým odporem, nemůže se chovat jako pasivní, tuhé kyvadlo. Kdyby to udělal, s každou další otáčkou by se jeho amplituda zmenšovala. Ztracenou energii musí do systému aktivně doplňovat **mechanickou prací svých svalů**.

Z fyzikálního a biomechanického hlediska tento proces využívá princip **parametrického buzení** (podobně jako když se člověk sám rozhoupává na houpačce). Klíčem je cílená změna polohy těžiště těla vůči ose otáčení (hrazdě) v různých fázích pohybu.

1. Fáze klesání (maximalizace poloměru)

* Při pohybu z nejvyššího bodu dolů se gymnasta snaží tělo co nejvíce "vytáhnout" z ramen a plně se narovnat.
* Tím posouvá své těžiště co nejdále od osy rotace.
* Zvětšením poloměru dráhy těžiště dochází k maximální možné přeměně gravitační potenciální energie na energii kinetickou. Moment setrvačnosti je v této fázi největší.

2. Fáze stoupání (práce proti odstředivé síle)

* Klíčový moment nastává ve spodní části oblouku a při následném stoupání. Gymnasta provede rychlý dynamický pohyb (tzv. *tap* – prohnutí a následné mírné "zabalení" v bocích a ramenou).
* Tímto pohybem efektivně **přitáhne své těžiště blíž k ose rotace**.
* Ve spodní části dráhy má gymnasta nejvyšší úhlovou rychlost a působí na něj maximální odstředivá síla směřující od hrazdy.

**Fyzikální podstata dodání energie**

Když gymnasta zkrátí poloměr rotace svého těžiště $r$ o vzdálenost $dr$ směrem k hrazdě, musí jeho svaly (především svalstvo pletence ramenního a středu těla) vykonat práci proti obrovské odstředivé síle $F_c$. Vykonaná mechanická práce $W$ je dána integrálem:

$$W = \int F_c \, dr$$

Tato pozitivní svalová práce se okamžitě transformuje do rotační energie systému. Můžeme se na to podívat i přes zmenšení momentu setrvačnosti $I$. Zkrácením těla se moment setrvačnosti zmenší, a aby zůstal zachován moment hybnosti $L$ (respektive reagoval na vnější momenty), musí se adekvátně zvýšit úhlová rychlost $\omega$:

$$L = I \cdot \omega$$

Ztráta energie třením je tedy beze zbytku (a často i s přebytkem pro zrychlení rotace) kompenzována vnitřní prací gymnasty při posunu těžiště proti setrvačným silám v dolní fázi obratu.

````



````{exercise} Energie sprintu
:label: ex-bezec

Běžec běží volně rychlostí 10 m/s. Při sprintu se jeho kinetická energie zvýší dvojnásobně. Urči rychlost běžce při sprintu.

```{figure} images/prx-007.png
:width: 550px
:align: center
:name: fig-bezec

Běžec na dráze.
````

````{solution} ex-bezec
:class: dropdown

**Zadání:**

* Počáteční rychlost: $v_1 = 10\text{ m/s}$
* Kinetická energie při sprintu: $E_{k2} = 2 \cdot E_{k1}$
* Rychlost při sprintu: $v_2 = {?}$


**Vztahy a odvození:**

1. Kinetická energie tělesa o hmotnosti $m$ a rychlosti $v$:

$$E_k = \frac{1}{2} m v^2$$


2. Vyjádření pro obě fáze běhu:

$$E_{k1} = \frac{1}{2} m v_1^2$$


$$E_{k2} = \frac{1}{2} m v_2^2$$


3. Dosazení podmínky ze zadání ($E_{k2} = 2 \cdot E_{k1}$):

$$\frac{1}{2} m v_2^2 = 2 \cdot \left(\frac{1}{2} m v_1^2\right)$$


4. Úprava rovnice (krácení faktorem $\frac{1}{2}m$):

$$v_2^2 = 2 \cdot v_1^2$$


5. Odmocnění:

$$v_2 = \sqrt{2} \cdot v_1$$


**Číselný výpočet:**


$$v_2 = \sqrt{2} \cdot 10\text{ m/s} \approx 1{,}414 \cdot 10\text{ m/s} \approx \mathbf{14{,}14\text{ m/s}}$$



Rychlost běžce při sprintu je přibližně **14,1 m/s** (respektive přesně $10\sqrt{2}\text{ m/s}$).

````
